### Part 2 :

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.linear_model import SGDRegressor
from sklearn.linear_model import LinearRegression
```

```python
# Separating target from train
x_train_full = train_df.drop(columns=["ID",'medv'])
y_train_full = train_df['medv']
```

```python
x_train_split,x_test_split,y_train_split,y_test_split = train_test_split(
    x_train_full,y_train_full,test_size=0.2,random_state=42
)
```

```python
numeric_features = train_df.drop(columns=["ID","rad",'chas','medv']).columns.tolist()
categorical_features = train_df.drop(columns=numeric_features).drop(columns=["ID",'medv']).columns.tolist()
print(categorical_features)
```

```python
numeric_transformers = StandardScaler()
categorical_transformers = OneHotEncoder(handle_unknown='ignore',drop='first')
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num',numeric_transformers,numeric_features),
        ('cat',categorical_transformers,categorical_features)
    ],
    remainder='passthrough'
)
```

```python
# Scale the split :
x_train_processed = preprocessor.fit_transform(x_train_split)
x_test_processed = preprocessor.transform(x_test_split)
```

```python
import numpy as np
```

```python
train_arr = np.c_[x_train_processed,np.array(y_train_split)]
train_arr[0]
```

```python
print(f"x train processed shape : {x_train_processed.shape}")
print(f"x test processed shape : {x_test_processed.shape}")
```

```python
sgd_reg = SGDRegressor(
    max_iter=1000,
    eta0=0.01,
    random_state=42
)
sgd_reg.fit(x_train_processed,y_train_split)
```

```python
# # y = mx + c
# sgd_reg.coef_ # gives the slope(m) of 20 feature
# sgd_reg.intercept_ # c
```

```python
linear_regression_ols = LinearRegression()
linear_regression_ols.fit(x_train_processed,y_train_split)
```

### Test Model :

```python
# SGDRegressor 
y_test_predict = sgd_reg.predict(x_test_processed)
y_train_predict = sgd_reg.predict(x_train_processed)
```

```python
#OLS (ordinary Least Square : for small to medium sized datasets and only 1 iteration means epoch is performed)
y_test_predict_ols = linear_regression_ols.predict(x_test_processed)
y_train_predict_ols = linear_regression_ols.predict(x_train_processed)
```

```python
def plot_actual_vs_predicted(y_actual,y_predict):
    if isinstance(y_actual,pd.Series):
        y_actual = y_actual.values

    plt.figure(figsize=(12,7))
    
    plt.plot(y_actual,label='Actual Values',linewidth=2,color='#004c6d')
    plt.plot(y_predict,label="Predicted Values",color='#ff7c43',linestyle='--',linewidth=2)

    plt.title('Actual vs. Predicted Values Over Index (Trend Comparison)',fontsize=18,fontweight='bold')
    plt.xlabel('Data Index')
    plt.ylabel('Median Home Value')
    plt.legend()
    plt.grid(True, linestyle=':', alpha=0.6)
    plt.tight_layout()
    plt.show()    
    
```

```python
plot_actual_vs_predicted(y_test_split,y_test_predict)
```

```python
plot_actual_vs_predicted(y_test_split,y_test_predict_ols)
# plot_actual_vs_predicted(y_train_split,y_train_predict_ols)
```

### Evaluation :

```python
from sklearn.metrics import mean_squared_error
```

```python
# SGD Regressor : 
mse_train = mean_squared_error(y_train_split,y_train_predict)
print(f"MSE (Train) : {mse_train}")

mse_test = mean_squared_error(y_test_split,y_test_predict)
print(f"MSE (Test) : {mse_test}")
```

- From the above evaluation we see it's an ideal fit (random_state = 42)

- Because MSE is squared, you can take the square root (RMSE) to put the error back into the original units of your data. The square root of 21.35 is roughly 4.62 . If you are predicting the Boston housing prices (where values are in $1,000s), your model's predictions are off by an average of about $4,620 on unseen data. If that margin of error is acceptable for your use case, your model is doing fine!  (random_state = 2)

```python
# OLS : Ordinary Least Squared -->
mse_train_ols = mean_squared_error(y_train_split,y_train_predict_ols)
print(f"MSE (Train) (OLS) : {mse_train_ols}")

mse_test_ols = mean_squared_error(y_test_split,y_test_predict_ols)
print(f"MSE (Test) (OLS) : {mse_test_ols}")
```

- From the above evaluation, we see it's an ideal fit, as the training MSE (21.61) and test MSE (22.38) are very close, indicating the model generalizes well without overfitting or underfitting.

- Because MSE is squared, you can take the square root (RMSE) to put the error back into the original units of your data. The square root of 22.38 is roughly 4.73. If you are predicting the Boston housing prices (where values are in $1,000s), your model's predictions are off by an average of about $4,730 on unseen data. If that margin of error is acceptable for your use case, your OLS model is performing well!

> <b>Model Comparison: The SGD Regressor performs better than the OLS model on unseen test data. The SGD Regressor achieved a lower test RMSE of 4.62 (an average error of $4,620) compared to OLS's test RMSE of 4.73 (an average error of $4,730), providing a tighter margin of prediction error. </b>
