```python
import joblib
import pandas as pd
```

```python
preprocessor_file_name = 'data_preprocessor.joblib'
model_file_name = 'final_sgd_reg_model.joblib'

loaded_model = joblib.load(model_file_name)
loaded_preprocessor = joblib.load(preprocessor_file_name)
```

```python
test_features_list = [
    50,    # crim
    0,     # zn
    18.1,  # indus
    0,     # chas
    0.7,   # nox
    4.0,   # rm 
    100,   # age
    1.0,   # dis 
    24,    # rad
    666,   # tax
    20.2,  # ptratio
    300,   # black
    30     # lstat 
]

# 13 columns
```

```python
Feature_Column = ['crim', 'zn', 'indus', 'chas', 'nox', 'rm', 'age', 'dis', 'rad','tax', 'ptratio', 'black', 'lstat']
```

```python
def make_prediction_from_list(input_list,feature_column,boston_model,preprocessor):
    df = pd.DataFrame([input_list],columns=feature_column)
    print(f"Input Dataframe : {df}\n")

    processed_df = preprocessor.transform(df)
    predictions =  boston_model.predict(processed_df)
    print(predictions,'\n')

    return f"{predictions[0]:.2f}"
```

```python
make_prediction_from_list(input_list=test_features_list,
                          feature_column=Feature_Column,
                          boston_model=loaded_model,preprocessor=loaded_preprocessor)
```
