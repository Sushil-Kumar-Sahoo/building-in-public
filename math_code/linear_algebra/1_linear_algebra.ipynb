* Bank robber has 150 km/h car and five-minute head start
* How long does it take the sheriff to catch the robber?
* What distance will they have traveled at that point?

```python
t = np.linspace(0,40,1000) # t represents time in minute
```

```python
# we know distance = speed * time 
# though speed here is given by km/hr so we need to convert them into km/min

# distance covered by robber at time ti : (150/60)* t = 2.5 * t
t_r = 2.5 * t               

# distance covered by Sheriff at time ti : (180/60) * (t-5) = 3 * (t-5)
t_s = 3 * (t-5)
```

># Norms

A **vector** represents both:

- **Magnitude (length)**
- **Direction** from the origin

For example,

$$
\mathbf{x} =
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
=
\begin{bmatrix}
12 \\
4
\end{bmatrix}
$$

or equivalently,

$$
[x_1, x_2] = [12, 4]
$$

The vector starts at the **origin** $(0,0)$ and points to the location $(12,4)$.

> **Norms** are functions that quantify the **magnitude (length)** of a vector.



### $L^2$ Norm (Euclidean Norm)

The **$L^2$ norm** of a vector is defined as

$$
\|\mathbf{x}\|_2 =
\sqrt{\sum_{i=1}^{n} x_i^2}
$$

where:

- $x_i$ is the $i^{th}$ element of the vector.
- $n$ is the number of elements (dimensions).

### Interpretation

- Measures the **Euclidean (straight-line) distance** from the origin to the vector.
- It is the **most commonly used norm** in machine learning.

### Notation

Instead of writing

$$
\|\mathbf{x}\|_2
$$

it is often written simply as

$$
\|\mathbf{x}\|
$$

when the Euclidean norm is implied.

---

## Example

For the vector

$$
\mathbf{x} =
\begin{bmatrix}
12 \\
4
\end{bmatrix}
$$

the $L^2$ norm is

$$
\|\mathbf{x}\|_2
=
\sqrt{12^2 + 4^2}
=
\sqrt{144 + 16}
=
\sqrt{160}
\approx 12.65
$$

This value represents the **distance from the origin** to the point $(12,4)$.

---

## Key Points

- A **norm** measures the **length (magnitude)** of a vector.
- The **$L^2$ norm** is also called the **Euclidean norm**.
- It measures the straight-line distance from the origin.
- It is the default norm used in many machine learning and deep learning algorithms.

```python
import math
```

```python
x = np.array([3,4,5])
 
# to calculate magnitude or distance from the origin (0,0,0) to (3,4,5) is 

# math.sqrt(math.pow(3,2)+math.pow(4,2)+math.pow(5,2))

def magnitude(arr:np.ndarray):
    sqr_sum = 0
    for i in x:
       sqr_sum += math.pow(i,2)
    return math.sqrt(sqr_sum)

magnitude(x)
```

```python
np.linalg.norm(x)           # built in 
```

>### Unit Vector :
A **Unit Vector** is a vector whose **magnitude (norm) is equal to 1**.
#### Definition :
A vector $\mathbf{x}$ is a unit vector if
$$
\boxed{\|\mathbf{x}\| = 1}
$$
#### Properties :
- Magnitude (length) = **1**
- Represents **direction only** 
#### Example :
For
$$
\mathbf{x}=
\begin{bmatrix}
0.6\\
0.8
\end{bmatrix}
$$

$$
\|\mathbf{x}\|
=
\sqrt{0.6^2+0.8^2}
=
1
$$

Therefore,

$$
\mathbf{x}
$$

is a **Unit Vector**.
#### Key Points :
- A Unit Vector is a vector with **unit norm**.
- Any non-zero vector can be converted into a unit vector by dividing it by its magnitude:
$$
\boxed{\hat{\mathbf{x}}=\frac{\mathbf{x}}{\|\mathbf{x}\|}}
$$

```python
#  Let's prove this : Any non-zero vector can be converted into a unit vector by dividing it by its magnitude
x = np.array([3,2,5])

print("Unit Vector : ",x,"\t                                It's Maginitude",np.linalg.norm(x))

unit_vector = x/magnitude(x)
print("Unit Vector : ",unit_vector,"\tIt's Maginitude",np.linalg.norm(unit_vector))
```

### $L^1$ Norm (Manhattan Norm) :

The **$L^1$ norm** of a vector is defined as

$$
\|\mathbf{x}\|_1 =
\sum_{i=1}^{n} |x_i|
$$

where:

- $x_i$ is the $i^{th}$ element of the vector.
- $|x_i|$ denotes the **absolute value** of $x_i$.
- $n$ is the number of elements (dimensions).

##### Interpretation :
- Another **commonly used norm** in machine learning.
- Varies **linearly** at all locations, whether near or far from the origin.
- Used whenever the **difference between zero and non-zero values is important**.
#### Geometric Interpretation :
- The **$L^1$ norm** measures the **Manhattan distance** (also called **Taxicab distance**) from the origin.
- Instead of taking the straight-line distance (Euclidean distance), it sums the absolute distances along each axis.
#### Key Points :
- The **$L^1$ norm** is also known as the **Manhattan norm** or **Taxicab norm**.
- It is calculated by **summing the absolute values** of all vector components.
- It treats each dimension independently and grows **linearly** with the magnitude of the vector.
- It is widely used in machine learning, especially in **L1 Regularization (Lasso Regression)**, where it encourages sparse models by driving some coefficients exactly to zero.

```python
x  = np.array([3, 4, 5])

def l_1_norm(arr:np.ndarray):
    sum = 0
    for i in arr:
        sum += np.abs(i)
    return sum

l_1_norm(x)
```

### Squared $L^2$ Norm :

The **Squared $L^2$ Norm** is the square of the Euclidean ($L^2$) norm. It measures the **squared magnitude (length)** of a vector.

#### Formula : 
$$
\boxed{\|\mathbf{x}\|_2^2=\sum_{i=1}^{n}x_i^2=\mathbf{x}^T\mathbf{x}}
$$
where:
- $\mathbf{x}$ is the vector.
- $\mathbf{x}^T$ is the transpose of $\mathbf{x}$.
#### Relation with $L^2$ Norm : 

| Norm | Formula |
|------|---------|
| $L^2$ Norm | $\sqrt{\sum_{i=1}^{n}x_i^2}$ |
| Squared $L^2$ Norm | $\sum_{i=1}^{n}x_i^2$ |

The Squared $L^2$ Norm simply removes the square root.

#### Example :
For

$$
\mathbf{x}=
\begin{bmatrix}
3\\
4
\end{bmatrix}
$$

$$
\|\mathbf{x}\|_2^2
=
3^2+4^2
=
25
$$

or

$$
\mathbf{x}^T\mathbf{x}
=
\begin{bmatrix}
3&4
\end{bmatrix}
\begin{bmatrix}
3\\
4
\end{bmatrix}
=
25
$$

---
#### Why is it used in Machine Learning?

- Computationally cheaper because it avoids the square root.
- Easy to differentiate, making optimization algorithms like **Gradient Descent** more efficient.
- Penalizes larger values more strongly than smaller values.

---

#### Limitation

The Squared $L^2$ Norm grows slowly near zero, so it **shrinks coefficients toward zero but usually does not make them exactly zero**.

---

#### Applications

- Mean Squared Error (MSE)
- Ridge Regression (L2 Regularization)
- Neural Networks
- Gradient Descent
- Weight Decay

---

#### Key Takeaways

- Measures the **squared length** of a vector.
- Formula:

$$
\boxed{\|\mathbf{x}\|_2^2=\sum_{i=1}^{n}x_i^2=\mathbf{x}^T\mathbf{x}}
$$

- Preferred in machine learning because it is computationally efficient and easy to optimize.

```python
# square L^2 Norm :
x = np.array([5,6,7])

# math.pow(np.linalg.norm(x),2)      # o/p : 110.0

x_transpose = x.T
# x_transpose @ x

np.dot(x_transpose,x)
```

* Returns the absolute value of the largest-magnitude element

### Generalized $L^p$ Norm

* Described by:

$$\|x\|_p = \left( \sum_i |x_i|^p \right)^{\frac{1}{p}}$$

* $p$ must be:
  * A real number
  * Greater than or equal to one
* Can derive $L^1$, $L^2$, and $L^\infty$ norm formulae by substituting for $p$
* Norms, particularly $L^1$ and $L^2$, used to regularize objective functions

>### The Dot Product :
If we have two vectors (say, $\mathbf{x}$ and $\mathbf{y}$) with the same length $n$, we can calculate the dot product between them. This is annotated several different ways, including the following:

* $x \cdot y$
* $x^T y$
* $\langle x, y \rangle$

Regardless which notation you use (I prefer the first), the calculation is the same; we calculate products in an element-wise fashion and then sum reductively across the products to a scalar value. That is, $x \cdot y = \sum_{i=1}^{n} x_i y_i$

The dot product is ubiquitous in deep learning: It is performed at every artificial neuron in a deep neural network, which may be made up of millions (or orders of magnitude more) of these neurons.
